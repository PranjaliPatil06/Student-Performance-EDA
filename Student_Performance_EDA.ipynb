# 1) What is the distribution of G3 scores in the Portuguese and Maths dataset?

```python
def analyze_g3_distribution_side_by_side(df_pre, df_mat):
    fig, axes = plt.subplots(1, 2, figsize=(14,5), sharey=True)

    # ---- Portuguese ----
    sns.histplot(df_pre['G3'], kde=True, bins=20, ax=axes[0], color='#4E79A7')
    axes[0].set_title("Portuguese – G3 Distribution with Density Curve", fontsize=12, weight='bold')
    axes[0].set_xlabel("Final Grade (G3)")
    axes[0].set_ylabel("Student Count")

    # ---- Math ----
    sns.histplot(df_mat['G3'], kde=True, bins=20, ax=axes[1], color='#F28E2B')
    axes[1].set_title("Math – G3 Distribution with Density Curve", fontsize=12, weight='bold')
    axes[1].set_xlabel("Final Grade (G3)")
    axes[1].set_ylabel("")

    plt.tight_layout()
    plt.show()

    # ---- Summary Stats ----
    print("\nPortuguese – G3 Summary Stats:")
    print(df_pre['G3'].describe())
    print(f"Skewness: {df_pre['G3'].skew():.3f}")

    print("\nMath – G3 Summary Stats:")
    print(df_mat['G3'].describe())
    print(f"Skewness: {df_mat['G3'].skew():.3f}")


# Run it
analyze_g3_distribution_side_by_side(df_pre, df_mat)
```

# 13) G1 - G2 - G3 relation for both the dataset 

```python
def driver_g1_side_by_side(df_pre, df_mat):
    fig, axes = plt.subplots(1, 2, figsize=(14,5), sharey=True)

    # --- Portuguese ---
    corr_pre = df_pre[['G1','G3']].corr().iloc[0,1]
    sns.regplot(ax=axes[0], x=df_pre['G1'], y=df_pre['G3'], color='#1f77b4')
    axes[0].set_title(f"Portuguese – G1 vs G3 (corr={corr_pre:.3f})", fontsize=12, weight='bold')
    axes[0].set_xlabel("G1 Score")
    axes[0].set_ylabel("G3 Score")

    # --- Math ---
    corr_mat = df_mat[['G1','G3']].corr().iloc[0,1]
    sns.regplot(ax=axes[1], x=df_mat['G1'], y=df_mat['G3'], color='#ff7f0e')
    axes[1].set_title(f"Math – G1 vs G3 (corr={corr_mat:.3f})", fontsize=12, weight='bold')
    axes[1].set_xlabel("G1 Score")
    axes[1].set_ylabel("")

    plt.tight_layout()
    plt.show()

# Call the combined function
driver_g1_side_by_side(df_pre, df_mat)
```

```python
def driver_g2_side_by_side(df_pre, df_mat):
    fig, axes = plt.subplots(1, 2, figsize=(14,5), sharey=True)

    # --- Portuguese ---
    corr_pre = df_pre[['G2','G3']].corr().iloc[0,1]
    sns.regplot(ax=axes[0], x=df_pre['G2'], y=df_pre['G3'], color='green')
    axes[0].set_title(f"Portuguese – G2 vs G3 (corr={corr_pre:.3f})", fontsize=12, weight='bold')
    axes[0].set_xlabel("G2 Score")
    axes[0].set_ylabel("G3 Score")

    # --- Math ---
    corr_mat = df_mat[['G2','G3']].corr().iloc[0,1]
    sns.regplot(ax=axes[1], x=df_mat['G2'], y=df_mat['G3'], color='green')
    axes[1].set_title(f"Math – G2 vs G3 (corr={corr_mat:.3f})", fontsize=12, weight='bold')
    axes[1].set_xlabel("G2 Score")
    axes[1].set_ylabel("")

    plt.tight_layout()
    plt.show()

# Call the function
driver_g2_side_by_side(df_pre, df_mat)
```

What we discovered

G1 gives an early hint about how a student will perform.

G2 almost fully determines G3 — students with high G2 end with high G3, and students with low G2 end with low G3.

🔹 What it means

The final grade (G3) is not a surprise.

By the time G2 is recorded, the student's outcome is already predictable.

G1 matters, but G2 is the real turning point.
